```python
import pandas as pd
import os

# Path to the downloaded dataset
path = "/kaggle/input/phishingemails"

# List all files inside the dataset folder to check the exact file name
files = os.listdir(path)
print("Files in dataset:", files)

# Assuming the dataset contains a CSV file, load the first CSV file found
csv_file = [f for f in files if f.endswith('.csv')][0]
file_path = os.path.join(path, csv_file)

# Read and display the first few rows of the dataset
df = pd.read_csv(file_path)
display(df.head())
```

### 1. Data Preprocessing and Cleaning
We will inspect the column names, drop any unnecessary columns, handle missing values, and map the targets to binary values (e.g., Phishing = 1, Safe = 0).

```python
import pandas as pd
import numpy as np

# Display columns and check for missing values
print("Dataset Columns:", df.columns)
print("Missing values:\n", df.isnull().sum())

# Drop unnamed index columns if they exist
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])

# Drop rows with null values in text or label
df = df.dropna(subset=['Email Text', 'Email Type'])

# Check class distributions
print("\nClass Distribution:\n", df['Email Type'].value_counts())

# Map Email Type to binary target (1 for Phishing, 0 for Safe)
# Standardize to avoid issues with extra whitespaces
df['Email Type'] = df['Email Type'].str.strip()
df['label'] = df['Email Type'].map({'Phishing Email': 1, 'Safe Email': 0})

# Filter out any unmapped rows just in case
df = df.dropna(subset=['label'])
df['label'] = df['label'].astype(int)

print("\nMapped Label Distribution:\n", df['label'].value_counts())
```

### 2. Feature Extraction (TF-IDF) and Train-Test Split
Next, we'll split the data into training and test sets, and vectorize the email texts using `TfidfVectorizer`.

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    df['Email Text'],
    df['label'],
    test_size=0.2,
    random_state=42,
    stratify=df['label']
)

# Initialize TF-IDF Vectorizer
vectorizer = TfidfVectorizer(stop_words='english', max_features=10000)

# Fit and transform the training text data, transform the test data
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print(f"Training set shape: {X_train_tfidf.shape}")
print(f"Testing set shape: {X_test_tfidf.shape}")
```

### 3. Model Training (Logistic Regression) & Evaluation
We will fit a Logistic Regression model and generate evaluation metrics.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Initialize and train the Logistic Regression Model
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train_tfidf, y_train)

# Predictions
y_pred = model.predict(X_test_tfidf)

# Metrics
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}\n")
print("Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Safe Email', 'Phishing Email']))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

# 1. Generate TF-IDF features for training and test sets using the fitted vectorizer
X_train_tfidf = vectorizer.transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# 2. Initialize and train the Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)
rf_model.fit(X_train_tfidf, y_train)

# 3. Predict on test data
y_pred_rf = rf_model.predict(X_test_tfidf)

# 4. Evaluate performance
rf_accuracy = accuracy_score(y_test, y_pred_rf)
print(f"Random Forest Accuracy: {rf_accuracy:.4f}\n")
print("Random Forest Classification Report:")
print(classification_report(y_test, y_pred_rf, target_names=['Safe Email', 'Phishing Email']))
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, roc_auc_score

# 1. Get the probability estimates for the positive class (Phishing Email = 1)
y_probs_lr = model.predict_proba(X_test_tfidf)[:, 1]
y_probs_rf = rf_model.predict_proba(X_test_tfidf)[:, 1]

# 2. Calculate ROC curve metrics
fpr_lr, tpr_lr, _ = roc_curve(y_test, y_probs_lr)
fpr_rf, tpr_rf, _ = roc_curve(y_test, y_probs_rf)

# 3. Calculate Area Under the Curve (AUC) scores
auc_lr = roc_auc_score(y_test, y_probs_lr)
auc_rf = roc_auc_score(y_test, y_probs_rf)

# 4. Plot the ROC curves
plt.figure(figsize=(8, 6))
plt.plot(fpr_lr, tpr_lr, color='blue', lw=2, label=f'Logistic Regression (AUC = {auc_lr:.4f})')
plt.plot(fpr_rf, tpr_rf, color='green', lw=2, label=f'Random Forest (AUC = {auc_rf:.4f})')
plt.plot([0, 1], [0, 1], color='grey', linestyle='--', lw=1.5, label='Random Guess')

# Add details and labels
plt.xlim([-0.02, 1.02])
plt.ylim([-0.02, 1.02])
plt.xlabel('False Positive Rate (1 - Specificity)', fontsize=12)
plt.ylabel('True Positive Rate (Sensitivity)', fontsize=12)
plt.title('ROC Curve Comparison', fontsize=14, fontweight='bold', pad=15)
plt.legend(loc='lower right', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 6. Packaging as a Flask API
We will create a Flask web API script. It will load our exported artifacts from `exports/` and define an endpoint `/predict` which accepts JSON payloads (e.g., `{"email_text": "your text here"}`) and returns the classification category and confidence score.

```python
flask_app_code = """
from flask import Flask, request, jsonify
import joblib
import os

app = Flask(__name__)

# Define paths to saved model artifacts
MODEL_PATH = 'exports/logistic_regression_model.pkl'
VECTORIZER_PATH = 'exports/tfidf_vectorizer.pkl'

# Load model and vectorizer once when server starts
if os.path.exists(MODEL_PATH) and os.path.exists(VECTORIZER_PATH):
    model = joblib.load(MODEL_PATH)
    vectorizer = joblib.load(VECTORIZER_PATH)
    print("Model and Vectorizer loaded successfully!")
else:
    raise FileNotFoundError("Model files not found. Please ensure they are in the exports/ directory.")

@app.route('/predict', methods=['POST'])
def predict():
    data = request.get_json(force=True)

    if not data or 'email_text' not in data:
        return jsonify({'error': 'Please provide email_text in request body.'}), 400

    email_text = data['email_text']

    # Perform prediction
    text_tfidf = vectorizer.transform([email_text])
    prediction = model.predict(text_tfidf)[0]
    probabilities = model.predict_proba(text_tfidf)[0]

    category = "Phishing Email" if prediction == 1 else "Safe Email"
    confidence = float(probabilities[prediction])

    return jsonify({
        'email_text': email_text,
        'prediction': category,
        'confidence': confidence
    })

if __name__ == '__main__':
    # Run on port 5000
    app.run(host='0.0.0.0', port=5000)
"""

# Save the script to a python file
with open('app.py', 'w') as f:
    f.write(flask_app_code)

print("Flask API application code written successfully to app.py!")
```
